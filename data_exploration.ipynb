```python
import pandas as pd
from etl.logging_config import log_step
import re
from pathlib import Path
```

```python


def clean(df, file_path):
    filename = Path(file_path).name
    match = re.match(r"yellow_tripdata_(\d{4})-(\d{2})\.parquet", filename) 

    target_year = int(match.group(1))

    with log_step("clean_trips", rows_in=len(df), year=target_year) as stats:
        cleaned = df.copy()
    # 1. keep only the valid date records

        cleaned["tpep_pickup_datetime"] = pd.to_datetime(
            cleaned["tpep_pickup_datetime"], errors="coerce"
        )
        cleaned["tpep_dropoff_datetime"] = pd.to_datetime(
            cleaned["tpep_dropoff_datetime"], errors="coerce"
        )
        
        valid_dates = (
                (cleaned["tpep_pickup_datetime"].dt.year == target_year) &
                (cleaned["tpep_dropoff_datetime"].dt.year == target_year) 
            )
        rows_before_date_filter = len(cleaned)
        cleaned = cleaned.loc[valid_dates].copy()
        stats["invalid_date_rows_removed"] = (
            rows_before_date_filter - len(cleaned)
        )
    
    # Fill missing values and record how many were filled
        fill_values = {
            "RatecodeID": 99,
            "passenger_count": 0,
            "store_and_fwd_flag": "Y",
            "congestion_surcharge": 0,
            "Airport_fee": 0,
        }

        for column, value in fill_values.items():
            stats[f"{column}_nulls_filled"] = int(cleaned[column].isna().sum())
            cleaned[column] = cleaned[column].fillna(value)

        stats["rows_out"] = len(cleaned)


    return cleaned
```

```python
file_path = Path("data/raw/yellow_tripdata_2026-07.parquet")
df_raw = pd.read_parquet(file_path)
cleaned_df = clean(df_raw, file_path)

print("Input rows:", len(df_raw))
print("Cleaned rows:", len(cleaned_df))
print("Rows removed:", len(df_raw) - len(cleaned_df))
print("Pickup years:", cleaned_df["tpep_pickup_datetime"].dt.year.unique())
print("Dropoff years:", cleaned_df["tpep_dropoff_datetime"].dt.year.unique())

assert cleaned_df["tpep_pickup_datetime"].dt.year.eq(2026).all()
assert cleaned_df["tpep_dropoff_datetime"].dt.year.eq(2026).all()
assert cleaned_df[list({
    "RatecodeID", "passenger_count", "store_and_fwd_flag",
    "congestion_surcharge", "Airport_fee"
})].isna().sum().sum() == 0
```
